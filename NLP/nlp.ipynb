```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("The cat chased the mouse.")
                
for token in doc:
      print(token.text, token.pos_, token.dep_, token.head.text)
```

```python
import nltk
import spacy

# Download the required tokenizer data package
nltk.download('punkt_tab')

text = "Natural Language Processing is interesting."

token = nltk.word_tokenize(text)
print(token)

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is opening a new office in London.")

for token in doc:
      print(token.text, token.pos_)
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
# words =["play", "playing", "played", "plays"]
words =["studies", "studying", "study"]
for word in words:
    print(word, "->", stemmer.stem(word))

```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = ["cars", "dogs", "running"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Ali works at Google in Lahore."

doc = nlp(text)

for entity in doc.ents:
    print(entity.text, entity.label_)
```

```python
from nltk.corpus import wordnet
synsets = wordnet.synsets("car")
for synset in synsets:
    print("1. ", synset)
    print("2. ", synset.name())
    print("3. ", synset.examples())
    print("4. ", synset.definition())
    print("---------------------------------------------------")
```

```python
# Remeove punctuation and special characters from a string

import string

text ="Hello, World! How are you?"

clean = text.translate(
str.maketrans('', '', string.punctuation)
)
print(clean)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
words = ["this", "is", "a", "very", "good", "movie"]

filtered = [
word for word in words
    if word.lower() not in stop_words
]
print(filtered) 
```

```python
import re

text1 = "I have 25 apples and 10 oranges."

text2 = "Contact me at user@example.com"

numbers = re.findall(r'\d+', text1)
print(numbers) 

email = re.findall(
r'[\w\.-]+@[\w\.-]+\.\w+', text2      
)

print(email)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("The big brown dog chased the small white cat.")

for chunk in doc.noun_chunks:
    print(chunk.text)   
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love Python",
    "I love Machine Learning",
    "Python is powerfull"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "machine learning is powerful",
    "machine learning is usefull"
]

vectorizer = CountVectorizer(
    ngram_range=(1,2)
)

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "I love machine learning",
    "Machine learning is amazing",
    "I like cooking food"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

similarity = cosine_similarity(X)

print(similarity)
```
