```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt 

sns.set_theme(style="whitegrid")

df = pd.DataFrame({
    "department": ["CS", "IT", "CS", "IT", "CS", "IT", "CS", "IT",],
    "gender": ["M", "F", "M", "F", "F", "M", "M", "F",],
    "marks": [85, 90, 78, 95, 88, 70, 92, 82,],
    "hours": [4, 6, 3, 7, 5, 2, 6, 4,],
    "age" : [20, 22, 21, 23, 24, 25, 22, 21,],
})

sns.scatterplot(
    data = df,
    x="hours",
    y="marks"
)
plt.show()
```

```python
sns.displot(df["marks"])
```

```python
sns.jointplot(
    data = df,
    x="hours",
    y="marks",
    kind= "scatter",
)
```

```python
sns.jointplot(
    data = df,
    x="hours",
    y="marks",
    kind= "kde",
)
```

```python
sns.jointplot(
    data = df,
    x="hours",
    y="marks",
    kind= "hex",
)
```

```python
sns.pairplot(
    df,
    vars=["marks", "hours", "age"],
    hue="department"
)
```

```python
sns.rugplot(
    data=df,
    x = "marks"
)
```

```python
sns.catplot(
    data=df,
    x= "department",
    y="marks",
    kind = "box"
)
```

```python
sns.boxplot(
    data=df,
    x= "department",
    y="marks",
)
```

```python
sns.violinplot(
    data=df,
    x= "department",
    y="marks",
)
```

```python
sns.stripplot(
    data=df,
    x= "department",
    y="marks",
    jitter = True,
)
```

```python
sns.swarmplot(
    data=df,
    x= "department",
    y="marks",
)
```

```python
sns.countplot(
    data=df,
    x= "department",
)
```
