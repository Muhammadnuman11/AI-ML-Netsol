```python
import pandas as pd
import matplotlib.pyplot as plt 

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 150, 130, 200, 250],
    "profit": [20, 35, 25, 50, 80],
})

df.plot(
    x="month",
    y="sales",
    kind="area"
)

plt.show()
```

```python
df.plot(
    x="month",
    y="sales",
    kind="line"
)
```

```python
df.plot(
    x="month",
    y="sales",
    kind="bar"
)
```

```python
df.plot(
    x="month",
    y="sales",
    kind="barh"
)
```

```python
df["sales"].plot(
    kind="hist",
    bins=5

)
```

```python
df["sales"].plot(kind="density")
```
