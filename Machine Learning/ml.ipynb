```python
import pandas as pd

df = pd.DataFrame({
    "area": [1000, 1500, None, 2500],
    "price": [8_000_000, 12_000_000, 16_000_000, 20_000_000]
})

df["area"] = df["area"].fillna(df["area"].median())

print(df)
```

```python
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000

m = len(X)

for epoch in range(epochs):

    y_pred = w * X + b
    errors = y_pred - y

    cost = np.sum(errors ** 2) / (2 * m)

    dw = np.sum(errors * X) / m
    db = np.sum(errors) / m

    w -= learning_rate * dw
    b -= learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Cost: {cost:.4f}")
print("Final weight:", w)
print("Final bias:", b)
```
